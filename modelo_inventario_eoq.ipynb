# **Modelo de Inventario EOQ - Caso de una Libreria**
Este cuaderno calcula la **Cantidad Economica de Pedido (EOQ)** para una libreria que vende cuadernos escolares. El objetivo es minimizar los costos de pedido y almacenamiento.

**EOQ (Economic Order Quantity o Cantidad Económica de Pedido)** es un modelo matemático que ayuda a las empresas a determinar cuánto pedir de un producto para minimizar los costos totales de inventario.

```python
import math
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

## 📋 **Parametros del Problema**

| Parametro | Simbolo | Valor | Unidad |
|-----------|---------|-------|--------|
| Demanda mensual | D | 500 | cuadernos/mes |
| Costo por pedido | S | RD$1,000 | pesos dominicanos |
| Costo de mantenimiento | H | RD$10 | pesos/unidad/mes |


### 🧮 **Formula EOQ**
$$ EOQ = \sqrt{\frac{2 \times D \times S}{H}} $$

Donde:
- **D**: Demanda por periodo
- **S**: Costo fijo por pedido
- **H**: Costo de mantenimiento por unidad por periodo

```python
# Parametros del problema
D = 500      # Demanda mensual (cuadernos)
S = 1000     # Costo de realizar un pedido (RD$)
H = 10       # Costo de mantenimiento por unidad por mes (RD$)

print(f"📈 Demanda mensual: {D} cuadernos")
print(f"💰 Costo por pedido: RD${S}")
print(f"🏗️  Costo de mantenimiento: RD${H}/unidad/mes")
print("-"*50)

# Calculo del EOQ
EOQ = math.sqrt((2 * D * S) / H)

print("🔍 RESULTADOS DEL ANALISIS EOQ:")
print(f"📦 Cantidad Economica de Pedido: {EOQ:.0f} cuadernos")
```
