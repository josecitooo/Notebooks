# **✅Punto 2**

- **Fuente**: Banco Central de la República Dominicana 
- **Datos Trimestrales de Exportaciones (2010–2024)**

## **📝 Contexto:**

Como parte de una estrategia nacional para fomentar las exportaciones, se ha solicitado a los científicos de datos del Ministerio de Industria, Comercio y MIPYMES que analicen el comportamiento estadístico de los sectores exportadores entre 2010 y 2024 para identificar tres sectores prioritarios que deben recibir mayores incentivos, asistencia técnica y facilidades de comercio exterior.

## **Los sectores a evaluar son:**

1.	**Minerales**
2.	**Agropecuarios (Nacionales y Zonas Francas)**
3.	**Industriales (Nacionales y Zonas Francas)**


- ***Cada sector incluye múltiples productos, pero para este ejercicio solo se requiere el análisis consolidado de cada sector a nivel agregado trimestral (puede usar datos simulados si no tiene acceso a las cifras reales por trimestre).***

### 🟣**Iniciemos importando los datos y librerias necesarias para el proyecto**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.stats import skew, kurtosis
```

### 🟣 **Ahora importamos nuestros datos**

```python
ruta = r"C:\Users\ferna\OneDrive - Instituto Tecnológico de Las Américas (ITLA)\data_tercer_parcial_ad2.csv"

df = pd.read_csv(ruta)
df.head()
```

## 🔍 **Requerimientos del análisis (valor 10 puntos):**


### 1.	**(2 puntos) Cálculo y presentación de las siguientes estadísticas descriptivas:**
- Mediana
- Moda
- Media
- Varianza
- Desviación estándar
- Coeficiente de variación
- Coeficiente de asimetría
- Coeficiente de curtosis
- Quintil 3 (percentil 75)

```python
datos = df['Cantidad de Exportaciones']


estadisticas = {
    'Estadística': [
        '🟢Media',
        '🟢Mediana',
        '🟢Moda',
        '🟢Varianza',
        '🟢Desviación estándar',
        '🟢Coeficiente de variación (%)',
        '🟢Coeficiente de asimetría',
        '🟢Coeficiente de curtosis',
        '🟢Quintil 3 (Percentil 75)'
    ],
    'Valor': [
        datos.mean(),
        datos.median(),
        datos.mode()[0] if not datos.mode().empty else "Sin valores modales únicos",
        datos.var(),
        datos.std(),
        (datos.std() / datos.mean()) * 100,
        skew(datos),
        kurtosis(datos),
        datos.quantile(0.75)
    ]
}

tabla = pd.DataFrame(estadisticas)

tabla
```

---
---

### 2.	**(2 puntos) Construcción e interpretación de un histograma por sector, que muestre la distribución de las exportaciones trimestrales.**

```python
# Voy a hacer subplots para hacer los 3 graficos en una sola figura, entonces voy a crear una figura de 18x6
plt.figure(figsize=(18, 6))

# Antes de hacer los histogramas, voy a crear una lista con los sectores y colores para cada uno de ellos
sectores = ['Agropecuarios', 'Industriales', 'Minerales']
colores = ['#4CAF50', '#2196F3', '#FFC107']

# Entonces por ultimo voy a hacer un for para recorrer los sectores y crear los histogramas
for i, sector in enumerate(sectores):
    plt.subplot(1, 3, i + 1)  # Posición del subplot
    datos_sector = df[df['Sector'] == sector]['Cantidad de Exportaciones']
    sns.histplot(datos_sector, bins=10, kde=True, color=colores[i], edgecolor='black')
    plt.title(f"Distribución de Exportaciones\n{sector}", fontsize=14)
    plt.xlabel("Cantidad de Exportaciones", fontsize=12)
    plt.ylabel("Frecuencia", fontsize=12 if i == 0 else 0)  # Solo mostrar etiqueta Y en el primer gráfico

# Ajustao el espacio entre gráficos
plt.tight_layout()


plt.show()
```

---
---

### 3.	**(2 puntos) Construcción e interpretación de un boxplot por sector. Analice presencia de valores atípicos, mediana y dispersión.**

```python
# Configurar el tamaño de la figura
plt.figure(figsize=(12, 8))

# Crear un boxplot por sector
sns.boxplot(data=df, x='Sector', y='Cantidad de Exportaciones', palette='Set2')

# Añadir título y etiquetas a los ejes
plt.title("Boxplot de Exportaciones Trimestrales por Sector", fontsize=16)
plt.xlabel("Sector", fontsize=14)
plt.ylabel("Cantidad de Exportaciones", fontsize=14)

# Mostrar el gráfico
plt.show()
```

---
---

### 4.	**(2 puntos) Comparación entre sectores a partir de los resultados anteriores:**
- ¿Qué sector presenta mayor crecimiento y cuál más estabilidad?
- ¿Cuál es más volátil?
- ¿Qué implicaciones tiene esto para la planificación de incentivos y políticas?

```python
# Agrupamos los datos por sector y calculamos las estadísticas
estadisticas = df.groupby('Sector')['Cantidad de Exportaciones'].agg(
    Media='mean',
    Desviacion_Estandar='std',
    Maximo='max',
    Minimo='min'
).reset_index()

# Creamos una figura con subplots
fig, axes = plt.subplots(1, 3, figsize=(20, 6), gridspec_kw={'width_ratios': [1, 1, 1.5]})

# Gráfico 1: Comparación de media (crecimiento)
sns.barplot(ax=axes[0], data=estadisticas, x='Sector', y='Media', palette='bright')
axes[0].set_title("Crecimiento por Sector\n(Media de Exportaciones)", fontsize=14)
axes[0].set_xlabel("Sector", fontsize=12)
axes[0].set_ylabel("Media de Exportaciones", fontsize=12)

# Gráfico 2: Comparación de desviación estándar (volatilidad)
sns.barplot(ax=axes[1], data=estadisticas, x='Sector', y='Desviacion_Estandar', palette='dark')
axes[1].set_title("Volatilidad por Sector\n(Desviación Estándar)", fontsize=14)
axes[1].set_xlabel("Sector", fontsize=12)
axes[1].set_ylabel("Desviación Estándar", fontsize=12)

# Gráfico 3: Gráfico combinado (rango de exportaciones)
for sector in estadisticas['Sector']:
    rango = estadisticas[estadisticas['Sector'] == sector]
    axes[2].plot([rango['Minimo'].values[0], rango['Maximo'].values[0]],
                 label=f"{sector}", marker='o')
axes[2].set_title("Estabilidad por Sector\n(Rango de Exportaciones)", fontsize=14)
axes[2].set_xlabel("Rango de Exportaciones", fontsize=12)
axes[2].set_ylabel("Valores Exportaciones", fontsize=12)
axes[2].legend(title="Sector")

plt.tight_layout()

plt.show()
```
