## 1. **Medidas de Tendencia Central**

- **Media aritmética**: Representa el promedio de los datos. Es sensible a valores extremos (outliers).
- **Mediana**: Es el valor central de un conjunto de datos ordenado. Es más robusta frente a la presencia de valores atípicos.
- **Moda**: Es el valor que ocurre con mayor frecuencia. 

### **Pregunta 1:**

- **¿En qué escenarios es más apropiado utilizar la mediana en lugar de la media? Fundamente su respuesta con un ejemplo.**

*La mediana es más apropiada en conjuntos de datos con **valores extremos o atípicos (outliers)**, ya que no se ve tan afectada por ellos. Por ejemplo, al analizar los ingresos anuales de una población, un pequeño grupo de individuos con ingresos extremadamente altos puede distorsionar la media, mientras que la mediana reflejará un ingreso más representativo.*

- **En el contexto de ciencia de datos, ¿cuándo sería preferible utilizar la moda en lugar de la media o la mediana?**

*La moda es útil cuando se analizan **datos categóricos** o **distribuciones discretas**. Por ejemplo, en una encuesta sobre el color favorito de una muestra de personas, la moda identifica el color más elegido. En datos numéricos, la moda puede destacar valores recurrentes importantes (e.g., tamaño de zapato más común en una población).*

---

## 2. **Medidas de Dispersión**

- **Rango**: Diferencia entre el valor máximo y el mínimo.
- **Varianza** y desviación estándar: Miden el grado de dispersión de los datos respecto a la media.
- **Rango** intercuartílico (IQR): Diferencia entre el tercer cuartil (Q3) y el primer cuartil (Q1); menos afectado por valores atípicos.

### **Pregunta 2:**

- **Si un conjunto de datos tiene desviación estándar igual a cero, ¿qué indica esto sobre los datos?**

*Una desviación estándar de cero significa que **todos los valores del conjunto de datos son iguales**, es decir, no hay variación.*


- **¿Qué consecuencias puede tener una varianza muy alta en el rendimiento de un modelo de aprendizaje automático?**

*Una varianza alta indica que los datos son muy dispersos, lo que puede llevar a **modelos poco robustos** o **sobreajuste (overfitting)**, ya que el modelo puede aprender ruido en lugar de patrones generalizables.*

- **¿Es siempre deseable una baja variabilidad en una variable predictora? Explique.**

*No necesariamente. Una baja variabilidad puede hacer que la variable predictora sea **poco relevante** para el modelo, ya que aporta poca información diferencial. Sin embargo, una variabilidad moderada es deseable para capturar patrones relevantes.*

- **En el tratamiento de valores faltantes (missing values), ¿qué medida estadística utilizaría (rango, IQR, desviación estándar) para decidir si una variable debe ser imputada o descartada? Justifique su elección.**

*El **IQR (rango intercuartílico)** es una buena opción, ya que es menos afectado por valores atípicos. Si el IQR de una variable es muy bajo (indicando poca variabilidad), podría ser más razonable descartar la variable en lugar de imputarla, ya que aporta poca información.*

---

## 3. **Medidas de Posición**

- **Cuartiles (Q1, Q2, Q3):** Dividen el conjunto de datos en cuatro partes iguales.
- **Percentiles**: Indican la posición relativa de un dato en la distribución, señalando el valor bajo el cual cae un determinado porcentaje de observaciones.

### **Pregunta 3:**

- **¿Qué relación existe entre el percentil 50 y la mediana? Explique con claridad.**

*El percentil 50 **es equivalente a la mediana**, ya que indica el punto en el que el 50% de las observaciones son menores o iguales a dicho valor. Ambos conceptos reflejan el valor central de la distribución.*

---

## 4. **Asimetría y Curtosis**


- **Positiva**: la cola de la distribución se extiende hacia la derecha.
- **Negativa**: la cola se extiende hacia la izquierda.
- **Curtosis**: Describe la forma del pico y el grosor de las colas de la distribución.
- **Leptocúrtica**: distribución con pico alto y colas gruesas (más probabilidad de valores extremos).
- **Platicúrtica**: distribución aplanada, con colas delgadas.

### **Pregunta 4:**

- **¿Qué implica una curtosis elevada respecto a la probabilidad de ocurrencia de valores extremos?**

*Una **curtosis elevada** indica una mayor probabilidad de ocurrencia de valores extremos en comparación con una distribución normal. Esto significa que las colas de la distribución son más gruesas.*

- **¿En qué situaciones una distribución leptocúrtica puede ser problemática al modelar?**

*En distribuciones leptocúrticas, los valores extremos pueden afectar negativamente la estabilidad de los modelos, causando sensibilidad excesiva a dichos puntos extremos. Esto puede generar predicciones menos fiables.*

- **¿Qué implicaciones tiene una asimetría negativa fuerte en una variable financiera, como el rendimiento de acciones? ¿Qué transformaciones estadísticas podrían aplicarse para corregir su asimetría?**

*Una asimetría negativa fuerte en rendimientos indica que las pérdidas grandes son más probables que las ganancias grandes. Esto es crucial para la gestión del riesgo. Para corregir esta asimetría, se pueden aplicar transformaciones como:*
  - **Raíz cuadrada o logaritmo:** *Reducen asimetría y comprimen valores extremos.*
  - **Box-Cox Transformation:** *Mejora la simetría de datos positivos.*

## 5. **Aplicaciones en Ciencia de Datos**

### **Pregunta 5:**

- **Al momento de entrenar un modelo, ¿cuál medida de dispersión (desviación estándar, IQR o rango) resulta más adecuada para decidir si una variable debe ser normalizada o escalada? Justifique su respuesta con base en criterios técnicos.**

*La **desviación estándar** es la medida más adecuada, ya que refleja cómo los datos están distribuidos alrededor de la media. Esto es fundamental para técnicas como **escalado estándar (StandardScaler)**, que transforma las variables para tener una media de 0 y desviación estándar de 1.*

### **Pregunta 6:**

- **¿Existe un umbral razonable de variabilidad por debajo del cual una variable puede considerarse irrelevante en un modelo predictivo? Fundamente su respuesta e indique cómo podría evaluarse dicha relevancia.**

*Sí, cuando la **varianza es cercana a cero**, la variable puede considerarse irrelevante, ya que no aporta información significativa para la predicción. Para evaluar esta relevancia, se pueden aplicar métodos como*:
  - **Análisis de varianza (ANOVA):** *Evalúa el impacto de la variable en la predicción.*
  - **Reducción de dimensionalidad (PCA):** *Identifica variables con baja contribución a la varianza total.*

### 🟢**Codigo**🟢

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import skew, kurtosis, mode
```

```python
df = pd.read_excel('Data Practica M3 AD2.xlsx')
df
```

```python
columnas_esperadas = [
    "INGRESOS TOTALES",
    "COSTOS Y GASTOS OPERACIONALES",
    "GASTO EN: COMBUSTIBLES",
    "GASTO EN: SERVICIOS BÁSICOS"
]

#🎯 Filtrar columnas que realmente estén en el DataFrame
columnas_presentes = [col for col in columnas_esperadas if col in df.columns]

#3🗃️ Mostrar las primeras filas de los datos
df[columnas_presentes].head(5)  # Puedes cambiar el número para ver más filas
```

## 🟢**Cálculo de medidas estadísticas**🟢

```python
# Inicializar un DataFrame para guardar los resultados
estadisticas = pd.DataFrame(columns=['Variable', 'Media', 'Mediana', 'Moda', 'Varianza', 'Desviación estándar', 'Coeficiente de variación', 'Coeficiente de asimetría', 'Coeficiente de curtosis', 'Quintil 3'])

for variable in columnas_presentes:
    datos = df[variable]
    moda = datos.mode()[0] if not datos.mode().empty else "Sin moda única"
    estadisticas = pd.concat([estadisticas, pd.DataFrame({
        'Variable': [variable],
        'Media': [datos.mean()],
        'Mediana': [datos.median()],
        'Moda': [moda],
        'Varianza': [datos.var()],
        'Desviación estándar': [datos.std()],
        'Coeficiente de variación': [(datos.std() / datos.mean()) * 100],
        'Coeficiente de asimetría': [skew(datos)],
        'Coeficiente de curtosis': [kurtosis(datos)],
        'Quintil 3': [datos.quantile(0.75)]
    })], ignore_index=True)

# Mostrar el DataFrame con las estadísticas
estadisticas.reset_index(drop=True, inplace=True)
estadisticas = estadisticas.round(2)  # Redondear a 2 decimales
estadisticas
```

## 🟢**Generación de gráficos**🟢

### **Histogramas**✅

```python
# Crear una figura con subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 10))  # 2 filas y 2 columnas de gráficos
fig.suptitle("Histogramas de Variables MIPYMES", fontsize=16)

# Iterar sobre las variables y los ejes
for ax, variable in zip(axes.flatten(), columnas_esperadas):
    datos = df[variable]
    sns.histplot(datos, bins=10, kde=True, color='skyblue', edgecolor='black', ax=ax)
    ax.set_title(variable, fontsize=12)
    ax.set_xlabel("Valores", fontsize=10)
    ax.set_ylabel("Frecuencia", fontsize=10)

# Ajustar el diseño para evitar solapamientos
plt.tight_layout(rect=[0, 0, 1, 0.96])  # Rect ajusta espacio para el título
plt.show()
```

## **Boxplots** ✅

```python
# Crear una figura con 2 filas y 2 columnas de subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 10))  # 2 filas y 2 columnas

# Iterar sobre las variables y los ejes
for ax, variable in zip(axes.flatten(), columnas_esperadas):
    datos = df[variable]
    sns.boxplot(x=datos, color='lightgreen', ax=ax)
    ax.set_title(f"{variable}", fontsize=12)
    ax.set_xlabel(variable, fontsize=10)

# Ajustar el diseño para evitar solapamientos
plt.tight_layout()
plt.show()
```
