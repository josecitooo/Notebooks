> # **Limpieza y Preparacion de Datos🧼**

```python
# Convertir fecha a datetime
if 'date' in df.columns:
    df['date'] = pd.to_datetime(df['date'])
    df = df.sort_values('date')
    
# ====================================
    
# Para seguir limpiando los datos, voy a convertir `promo` y `holiday` a tipo booleano

# Promo
if 'promo' in df.columns:
    df['promo'] = df['promo'].astype(bool)
    
# Holiday    
if 'holiday' in df.columns:
    df['holiday'] = df['holiday'].astype(bool)
    
# ====================================

# Impresion de los tipos de datos
print("\n🟢Tipos de datos de cada columna🟢:")
df.dtypes
```

> # **Modelo de prediccion🔮**

- He estado aprendiendo a entrenar modelos con pipelines, entonces voy a aprovechar para incorporar el conociemiento aqui
- Un pipeline me va ayudar a organizar y automatizar todo el proceso de entrenamiento del modelo, asi evito los errores de mi parte

## **Preparacion de caracterisitcas 1️⃣**

```python
# Creamos primero las caracteristicas temporales simples
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day_of_week'] = df['date'].dt.dayofweek
df['is_weekend'] = (df['day_of_week'] >= 5).astype(int) # is_weekend es 1 si es sabado o domingo, 0 si es lunes a viernes

# Convertimos otra vez de booleanos a enteros para el modelo, esto es opcional pero puede ser util para algunos modelos... mejor prevenir 
df['promo_int'] = df['promo'].astype(int)
df['holiday_int'] = df['holiday'].astype(int)

# Seleccionamos las caracteristicas para el modelo y la variable objetivo

features = ['store', 'year', 'month', 'day_of_week', 'is_weekend', 'promo_int', 'holiday_int']
X = df[features] # Las caracteristicas que vamos a usar para predecir las ventas
y = df['sales'] # La variable objetivo, que es la cantidad de ventas que queremos predecir

print("🟢Caracteristicas seleccionadas🟢:")
features
```

    border: none;
    border-top: 15px solid #003366;

## **Division de los datos 2️⃣**

```python
# Dividimos los datos en entrenamiento y prueba (75% entrenamiento, 25% prueba)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=356) 

print(f"Entrenamiento: {X_train.shape}")
print(f"Prueba: {X_test.shape}")
```

<hr style="
    border: none;
    border-top: 15px solid #003366;
    border-radius: 8px;
    width: 80%;
    margin: 80px auto;
">

## **Creacion del Pipeline 3️⃣**

```python
# Pipeline con escalador y Random Forest... 
# Pude usar GridSearchCV para optimizar los hiperparametros, pero por simplicidad no lo hice
pipeline = Pipeline([
    ('scaler', StandardScaler()), # Voy a usar un escalador para normalizar las caracteristicas
    ('model', RandomForestRegressor(n_estimators=100, random_state=356)) # Modelo Random Forest con 100 arboles
])

print("🟢Entrenando modelo🟢:")
pipeline.fit(X_train, y_train)
```

<hr style="
    border: none;
    border-top: 15px solid #003366;
    border-radius: 8px;
    width: 80%;
    margin: 80px auto;
">

## **Predicciones y metricas 4️⃣**

```python
# predicciones
y_pred = pipeline.predict(X_test)

# Metricas
mae = mean_absolute_error(y_test, y_pred) # Error absoluto medio
rmse = np.sqrt(mean_squared_error(y_test, y_pred)) # Raiz del error cuadratico medio
r2 = r2_score(y_test, y_pred) # Coeficiente de determinacion R^2, que mide la calidad del ajuste del modelo
mape = np.mean(np.abs((y_test - y_pred) / y_test)) * 100  # Error porcentual
accuracy = 100 - mape # La presicion va a ser 100 menos el error porcentual

print("\n🟢Resultados del modelo🟢:")
print(f"Error absoluto medio: ${mae:.2f}")
print(f"Raiz del error cuadratico medio: ${rmse:.2f}")
print(f"R2: {r2:.3f}")
print(f"Error porcentual: {mape:.2f}%")
print(f"Precision: {accuracy:.2f}%") # Vuelvo y digo, la presicion es 100 menos el error porcentual
```

<hr style="
    border: none;
    border-top: 15px solid #003366;

">

## **Verificacion para ver si hay Overfitting 5️⃣**

```python
# Metricas en entrenamiento para comparar
y_pred_train = pipeline.predict(X_train) # Predicciones en el conjunto de entrenamiento
r2_train = r2_score(y_train, y_pred_train) # Coeficiente de determinacion R^2 en el conjunto de entrenamiento
mae_train = mean_absolute_error(y_train, y_pred_train) # Error absoluto medio en el conjunto de entrenamiento

print(f"\n🟢Verificacion de Overfitting🟢:")
print(f"R2 Entrenamiento: {r2_train:.3f}")
print(f"R2 Prueba: {r2:.3f}")
print(f"Diferencia: {abs(r2_train - r2):.3f}")

if abs(r2_train - r2) < 0.05: # Si la diferencia entre R2 de entrenamiento y prueba es menor a 0.05, no hay overfitting
    print("✅ NO HAY OVERFITTING")
else:
    print("⚠️ POSIBLE OVERFITTING")
```

### **📈 Ventas Promedio por Mes**

```python
%%R
# Ya sabemos TODAS las ventas por tienda, ahora vamos a ver las ventas promedio por mes
df %>%
  group_by(month) %>%
  summarise(avg_sales = mean(sales), .groups = 'drop') %>% # Agrupamos por mes y calculamos la venta promedio
  
  ggplot(aes(x = factor(month), y = avg_sales)) + # Graficamos el mes como factor y las ventas promedio
  
  geom_col(fill = "steelblue", alpha = 0.8) + # Columna azul para las ventas promedio
  
  # Añadimos etiquetas con el valor de las ventas promedio encima de cada barra (para que se vea mejor)
  geom_text(aes(label = paste0("$", round(avg_sales, 0))), 
            vjust = -0.5, size = 3) +
  
  # Añadimos titulo y etiquetas a los ejes
  labs(title = "Ventas Promedio por Mes", 
       x = "Mes", y = "Ventas Promedio ($)") + 
  
  # Al titulo le damos un tamaño y lo centramos con hjust 0.5 
  theme(plot.title = element_text(size = 16, hjust = 0.5))
```

### **🎯 Impacto de Promociones y Días Festivos**

```python
%%R
# Efecto de promociones y días festivos
df %>%
     # Agrupamos por promo y holiday, calculamos la venta promedio
  mutate(promo_label = ifelse(promo, "Con Promo", "Sin Promo"),
         holiday_label = ifelse(holiday, "Festivo", "Normal")) %>%
  
  # El group_by lo hacemnos por las etiquetas de promo y holiday
  group_by(promo_label, holiday_label) %>%
  
  # Calculamos la venta promedio por cada combinacion de promo y holiday con summarise y mean
  summarise(avg_sales = mean(sales), .groups = 'drop') %>%
  
  # Graficamos la interaccion entre promo y holiday para ver su efecto en las ventas
  ggplot(aes(x = interaction(promo_label, holiday_label), 
             y = avg_sales, 
             fill = interaction(promo_label, holiday_label))) +
  
  # Usamos geom_col para crear las barras o el grafico de barras
  geom_col() +
  
  # Añadimos etiquetas con el valor de las ventas promedio encima de cada barra (para que se vea mejor)
  geom_text(aes(label = paste0("$", round(avg_sales, 0))), 
            vjust = -0.5, size = 3) + # vjust -0.5 para que las etiquetas queden encima de las barras
  
  # Añadimos etiquetas y tema 
  labs(title = "Impacto de Promociones y Días Festivos", 
       x = "Condicion", y = "Ventas Promedio ($)") +
  theme(axis.text.x = element_text(angle = 45, hjust = 1),
        legend.position = "none",
        plot.title = element_text(size = 16, hjust = 0.5))
```
