**Contexto:**

> **👉 Se han registrado incidentes técnicos por cada embarcación durante los últimos 6 meses. Algunos eventos pueden modelarse con Poisson (número de allos técnicos), Binomial (embarques exitosos) o Hipergeométrica (selección aleatoria de tripulantes para control antidopaje).**

> **💡 ¿Cuál es la probabilidad de que ocurran más de 3 incidentes en un día? ¿Qué medidas deben tomarse en logística y prevención?**


```python
# Renombramos las columnas para facilitar la lectura
# de los datos en el caso 3, ya que las columnas tienen nombres largos y complicados
# Esto es para que el codigo sea mas legible y entendible

datos3.rename(columns={
    'FallosTecnicos (poisson)': 'Fallos',
    'EmbarquesExitosos (Binomial)': 'Exitos',
    'TripulantesControlados (Hiper)': 'Controlados'
}, inplace=True)

# ===============================
# 1️⃣ Estadisticas descriptivas
# ===============================

def mostrar_estadisticas(serie, nombre):
    print(f"\n📌 {nombre.upper()}")
    print(f"Media: {serie.mean():.2f}")
    print(f"Mediana: {serie.median():.2f}")
    print(f"Moda: {serie.mode().iloc[0] if not serie.mode().empty else 'No unica'}")
    print(f"Desviacion estandar: {serie.std():.2f}")
    print(f"Varianza: {serie.var():.2f}")
    print(f"Rango: {serie.min()} - {serie.max()}")

print("📊 CASO 3: INCIDENTES EN EMBARCACIONES DE CARGA")
print("="*60)
print("\n1️⃣ ESTADISTICAS DESCRIPTIVAS")
mostrar_estadisticas(datos3['Fallos'], "Fallos Tecnicos (Poisson)")
mostrar_estadisticas(datos3['Exitos'], "Embarques Exitosos (Binomial)")
mostrar_estadisticas(datos3['Controlados'], "Tripulantes Controlados (Hipergeometrica)")

# ===============================
# 2️⃣ Visualizacion con graficos mas grandes
# ===============================

print("\n2️⃣ VISUALIZACION CON GRAFICOS ")

# Aumentamos el tamaño de la figura principal (22,12)
plt.figure(figsize=(22, 12))  # Figura mas grande
variables = ['Fallos', 'Exitos', 'Controlados']
colores = ['skyblue', 'lightgreen', 'salmon']

# Lo hacemos con un bucle para evitar repetir codigo
# y asi hacer el codigo mas limpio y facil de entender

for i, var in enumerate(variables):
    plt.subplot(2, 3, i+1)
    sns.histplot(datos3[var], kde=True, bins=range(datos3[var].min(), datos3[var].max()+2), 
                color=colores[i], line_kws={'linewidth': 3})
    plt.title(f'Histograma - {var}', fontsize=16)  # Titulo
    plt.xlabel(var, fontsize=14)  # Etiqueta X
    plt.ylabel('Frecuencia', fontsize=14)  # Etiqueta Y
    plt.xticks(fontsize=12)  # Numeros del eje X
    plt.yticks(fontsize=12)  # Numeros del eje Y
    plt.grid(True, alpha=0.3)
    
    # Diagramas de caja 
    plt.subplot(2, 3, i+4)
    sns.boxplot(y=datos3[var], color=colores[i], width=0.5)
    plt.title(f'Diagrama de Caja - {var}', fontsize=16)  # Titulo
    plt.ylabel(var, fontsize=14)  # Etiqueta Y 
    plt.yticks(fontsize=12)  # Numeros del eje Y
    plt.grid(True, alpha=0.3)

# Titulo principal
plt.suptitle("Visualizacion de Variables - Incidentes Operativos", fontsize=20, fontweight='bold')
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()


# ===============================
# 3️⃣ Aplicacion de modelos estadisticos segun la columna
# ===============================

# --- POISSON: Probabilidad de que ocurran mas de 3 fallos tecnicos en un dia ---
media_fallos = datos3['Fallos'].mean()
# P(X > 3) = 1 - P(X <= 3)
prob_mas_3_fallos = 1 - poisson.cdf(3, mu=media_fallos)
print(f"\n📍 Poisson - Probabilidad de >3 fallos tecnicos en un dia: {prob_mas_3_fallos:.4f} ({prob_mas_3_fallos*100:.2f}%)")

# --- BINOMIAL: Probabilidad estimada de exito en embarques ---
# Suponiendo n=10 intentos por registro (segun los datos)
n_binomial = 10
p_exito = datos3['Exitos'].mean() / n_binomial
print(f"📍 Binomial - Probabilidad estimada de exito por embarque: {p_exito:.4f} ({p_exito*100:.2f}%)")

# --- HIPERGEOMeTRICA: Probabilidad de detectar al menos 1 controlado si hay 10% dopados ---
# Supongamos poblacion N=30, K=3 dopados (10%), n=muestra promedio
N = 30
K = 3
n_hiper = int(datos3['Controlados'].mean())
# P(X >= 1) = 1 - P(X = 0)
prob_al_menos_1 = 1 - hypergeom.pmf(0, N, K, n_hiper)
print(f"📍 Hipergeometrica - Prob. detectar al menos 1 dopado : {prob_al_menos_1:.4f} ({prob_al_menos_1*100:.2f}%)")
```

```python
print("📊 CASO 4: EVALUACION DE PRECISION DE DISPAROS EN ENTRENAMIENTO")
print("="*70)

# ================================
# 1️⃣ Analisis exploratorio de datos
# ================================

print("\n1️⃣ ANALISIS EXPLORATORIO DE DATOS")
print("-" * 40)
print(f"Registros totales: {len(datos4)}")
print(f"Primeras 5 observaciones:")
print(datos4.head())

# *********************************************************************************

# ================================
# 2️⃣ Medidas descriptivas
# ================================

print("\n2️⃣ ESTADISTICAS DESCRIPTIVAS")
print("-" * 40)
media = datos4['Total_Score_10Shots'].mean()
mediana = datos4['Total_Score_10Shots'].median()
moda = datos4['Total_Score_10Shots'].mode()
desviacion = datos4['Total_Score_10Shots'].std()

print(f"Media: {media:.2f} puntos")
print(f"Mediana: {mediana:.2f} puntos")
print(f"Moda: {moda.iloc[0]:.2f} puntos")
print(f"Desviacion estandar: {desviacion:.2f} puntos")
print(f"Rango: {datos4['Total_Score_10Shots'].min()} - {datos4['Total_Score_10Shots'].max()} puntos")

# *********************************************************************************

# ================================
# 4️⃣ Visualizacion de datos
# ================================

print("\n3️⃣ REPRESENTACION GRAFICA")
print("-" * 40)

# Histograma de los puntajes totales, este lo haremos con seaborn para que se vea mas bonito
plt.figure(figsize=(10, 5))
sns.histplot(datos4['Total_Score_10Shots'], kde=True, bins=20, color='skyblue')
plt.title('Histograma de Puntajes Totales')
plt.xlabel('Puntaje Total (10 tiros)')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.show()

# Diagrama de caja
plt.figure(figsize=(8, 3))
sns.boxplot(x=datos4['Total_Score_10Shots'], color='lightgreen')
plt.title('Diagrama de Caja del Puntaje Total')
plt.xlabel('Puntaje Total')
plt.show()

# *********************************************************************************

# ================================
# 4️⃣ Aplicacion del Teorema del Limite Central
# ================================

print("\n4️⃣ TEOREMA DEL LIMITE CENTRAL")
print("-" * 40)
n = len(datos4)
n_grupos = n // 30  # numero de grupos completos de 30 soldados
grupos_reales = [datos4['Total_Score_10Shots'][i*30:(i+1)*30] for i in range(n_grupos)]
medias_reales = [grupo.mean() for grupo in grupos_reales]

print(f"Numero total de soldados: {n}")
print(f"Grupos de 30 soldados: {n_grupos}")
print(f"Media poblacional: {media:.2f}")
print(f"Media de las medias muestrales: {np.mean(medias_reales):.2f}")
print(f"Desviacion estandar de las medias muestrales: {np.std(medias_reales):.2f}")
print(f"Error estandar teorico: {desviacion/np.sqrt(30):.2f}")

# Graficamos la distribucion de las medias reales
plt.figure(figsize=(10, 5))
sns.histplot(medias_reales, kde=True, color='orange')
plt.title('Distribucion Real de Medias Muestrales (n=30)')
plt.xlabel('Media de Muestra')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.axvline(media, color='red', linestyle='--', label=f'Media poblacional: {media:.2f}')
plt.axvline(np.mean(medias_reales), color='blue', linestyle='--', 
           label=f'Media de las muestras: {np.mean(medias_reales):.2f}')
plt.legend()
plt.show()
```
