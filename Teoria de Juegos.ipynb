```python
import nashpy as nash
import numpy as np

# Definimos las matrices de pagos para ambos supermercados
A = np.array([
    [70, 30],  # A: Precio alto
    [90, 50]   # A: Precio bajo
])
B = np.array([
    [70, 90],  # B: Precio alto
    [30, 50]   # B: Precio bajo
])

# Creamos el juego
game = nash.Game(A, B)

# Equilibrios de Nash
print("Equilibrios de Nash (puros y mixtos):")
for eq in game.support_enumeration():
    estrategia_A = "Precio alto" if np.allclose(eq[0], [1, 0]) else "Precio bajo"
    estrategia_B = "Precio alto" if np.allclose(eq[1], [1, 0]) else "Precio bajo"
    print(f"({estrategia_A}, {estrategia_B})")

# Resultados Pareto eficientes
print("\nResultados Pareto eficientes:")
acciones = ["Precio alto", "Precio bajo"]
outcomes = [(i, j, A[i, j], B[i, j]) for i in range(2) for j in range(2)]
pareto = []
for i, j, a, b in outcomes:
    dominated = False
    for i2, j2, a2, b2 in outcomes:
        if (a2 >= a and b2 >= b) and (a2 > a or b2 > b):
            dominated = True
            break
    if not dominated:
        pareto.append((acciones[i], acciones[j], a, b))
for res in pareto:
    print(f"(A: {res[0]}, B: {res[1]}) = ({res[2]}, {res[3]})")
```

### 1. ¿Hay algún equilibrio de Nash?

Sí, el equilibrio de Nash ocurre cuando ambos supermercados eligen **precio bajo**. En este caso, ninguno puede mejorar su resultado cambiando unilateralmente su estrategia:

- Si A elige precio bajo, B también prefiere precio bajo (50 > 30).
- Si B elige precio bajo, A también prefiere precio bajo (50 > 30).

Por lo tanto, (**precio bajo**, **precio bajo**) = **(50, 50)** es el equilibrio de Nash.

---

### 2. ¿Qué estrategia debería adoptar cada supermercado?

Cada supermercado debería elegir **precio bajo** para maximizar su propio beneficio, ya que es la mejor respuesta a la estrategia del otro.

---

### 3. ¿Cuál o cuáles resultados son óptimos de Pareto?

Un resultado es óptimo de Pareto si no se puede mejorar el resultado de un jugador sin empeorar el del otro. En este caso, el resultado (**precio alto**, **precio alto**) = **(70, 70)** es óptimo de Pareto, ya que ambos obtienen un mayor beneficio que en el equilibrio de Nash, pero requiere cooperación.

---

**Resumen:**
- **Equilibrio de Nash:** (precio bajo, precio bajo) = (50, 50)
- **Óptimo de Pareto:** (precio alto, precio alto) = (70, 70)

# Resolución del Ejercicio 2

### 1. Equilibrios de Nash

Analizando las estrategias posibles:

|                   | Y: Invierte | Y: No invierte |
|-------------------|:-----------:|:--------------:|
| **X: Invierte**   |  (50, 50)   |   (90, 10)     |
| **X: No invierte**|  (10, 90)   |   (60, 60)     |

- Si Y invierte, X prefiere invertir (50 > 10).
- Si Y no invierte, X prefiere invertir (90 > 60).
- Si X invierte, Y prefiere invertir (50 > 10).
- Si X no invierte, Y prefiere invertir (90 > 60).

Por lo tanto, el único equilibrio de Nash es (**Invierte**, **Invierte**) = **(50, 50)**.

---

### 2. ¿Existe alguna estrategia dominante?

Sí, para ambos jugadores la estrategia dominante es **invertir** en publicidad, ya que siempre les da un mejor resultado sin importar lo que haga la otra empresa.

### 3. Resultado más justo según Pareto eficiencia

Un resultado es Pareto eficiente si no se puede mejorar el resultado de un jugador sin empeorar el del otro.  
Los resultados (60, 60) y (50, 50) son Pareto eficientes, pero **(60, 60)** (ambos no invierten) es el más justo, ya que ambos obtienen el mismo beneficio y es el mayor posible para ambos simultáneamente.

---

**Resumen:**
- **Equilibrio de Nash:** (Invierte, Invierte) = (50, 50)
- **Estrategia dominante:** Invertir para ambos
- **Resultado más justo (Pareto eficiente):** (No invierte, No invierte) = (60, 60)
