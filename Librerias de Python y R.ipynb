# **5. Scikit-learn**

> **Scikit-learn es una librería de Python diseñada para el aprendizaje automático. Proporciona herramientas para tareas como clasificación, regresión, clustering, reducción de dimensionalidad, y más. Es ampliamente utilizada en ciencia de datos y machine learning debido a su simplicidad y eficiencia.**

```python
from sklearn.datasets import load_iris # Cargamos el dataset Iris de scikit-learn para realizar un ejemplo de clasificación
from sklearn.model_selection import train_test_split # Importamos la función para dividir el dataset en conjunto de entrenamiento y conjunto de prueba
from sklearn.ensemble import RandomForestClassifier # Importamos el clasificador Random Forest de scikit-learn para usarlo de ejemplo
from sklearn.metrics import accuracy_score # Para evaluar el modelo, importamos la métrica de exactitud

# Cargamos el dataset Iris
data = load_iris()
X = data.data
y = data.target

# Uno de los primeros pasos es dividir el dataset en conjunto de entrenamiento y conjunto de prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42) # Le hemos dado un 30% de los datos para el conjunto de prueba

# Creamos y entrenamos un modelo de Random Forest simple 
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Entonces ahora podemos hacer predicciones
y_pred = model.predict(X_test)

# Evaluamos el modelo
print("Exactitud del modelo:", accuracy_score(y_test, y_pred))
```

# **6. Statsmodels**

> **Statsmodels es una librería de Python para realizar análisis estadístico. Proporciona herramientas para construir modelos estadísticos como regresión lineal, ARIMA, pruebas de hipótesis, entre otros. Es ampliamente utilizada en econometría y análisis de datos.**

```python
import statsmodels.api as sm
import numpy as np

# Datos de ejemplo
X = np.array([1, 2, 3, 4, 5])  # Variable independiente
y = np.array([2, 4, 6, 8, 10])  # Variable dependiente

# Agregamos una constante para el modelo
X = sm.add_constant(X)

# Creamos el modelo de regresión lineal
model = sm.OLS(y, X).fit() # .OLS significa "Ordinary Least Squares" (Mínimos Cuadrados Ordinarios), que es un método común para ajustar modelos de regresión lineal.

# Resumen del modelo
print(model.summary()) 
```

# **5. Shiny**

> **`shiny` es un paquete de R que permite crear aplicaciones web interactivas directamente desde R. Es ideal para construir dashboards y reportes dinámicos sin necesidad de conocimientos avanzados en desarrollo web.**

```python
%%R

library(shiny)

# Definimos la interfaz de usuario
ui <- fluidPage(
  titlePanel("Ejemplo básico de Shiny"),
  sidebarLayout(
    sidebarPanel(
      sliderInput("num", "Selecciona un número:", min = 1, max = 100, value = 50)
    ),
    mainPanel(
      textOutput("resultado")
    )
  )
)

# Definimos el servidor
server <- function(input, output) {
  output$resultado <- renderText({
    paste("El número seleccionado es:", input$num)
  })
}

# Ejecutamos la aplicación
shinyApp(ui = ui, server)
```

```python
%%R

library(lubridate)

# Creamos una fecha de ejemplo
fecha <- "2025-07-09"

# Convertimos la cadena a un objeto de fecha
fecha_obj <- ymd(fecha)

# Extraemos componentes
anio <- year(fecha_obj)
mes <- month(fecha_obj)
dia <- day(fecha_obj)

# Mostramos los resultados
print(paste("Anio:", anio))
print(paste("Mes:", mes))
print(paste("Dia:", dia))
```

> **`Stringr` es un paquete de R diseñado para trabajar con cadenas de texto de manera eficiente. Proporciona funciones intuitivas para manipular, buscar, reemplazar y analizar texto.**

```python
%%R

library(stringr)

# Cadena de ejemplo
texto <- "Hola profe soy Jose espero que este bien"

# Operaciones con stringr
longitud <- str_length(texto)  # Longitud de la cadena
palabras <- str_split(texto, " ")  # Dividir la cadena en palabras
nuevo_texto <- str_replace(texto, "R", "programación")  # Reemplazar texto

# Mostrar resultados
print(paste("Longitud:", longitud))
print(paste("Palabras:", paste(unlist(palabras), collapse = ", ")))
print(paste("Nuevo texto:", nuevo_texto))
```

# **10. sf**

> **`sf` es un paquete de R diseñado para trabajar con datos espaciales. Permite manipular, analizar y visualizar datos geográficos como puntos, líneas y polígonos. Es ideal para proyectos relacionados con mapas y análisis geoespacial.**

```python
%%R

library(sf)

# Creamos un conjunto de datos espaciales de ejemplo
puntos <- st_as_sf(data.frame(
  id = 1:3,
  x = c(-70.5, -70.6, -70.7),
  y = c(19.5, 19.6, 19.7)
), coords = c("x", "y"), crs = 4326)

# Mostramos los datos espaciales
print(puntos)

# Calculamos la distancia entre los puntos
distancias <- st_distance(puntos)
print(distancias)
```
