# **Caso 2️⃣ – competencia estratégica entre empresas de transporte urbano**

**📊 Datos del Problema - Teoría de Juegos**

 **🚌 Empresas Competidoras**
- **TransCaribe**: Empresa de transporte urbano 1
- **MetroUrbano**: Empresa de transporte urbano 2

 **🎯 Estrategias Disponibles**
- **Estrategia A**: Renovar flota + subida moderada de tarifas
- **Estrategia B**: Mantener flota actual + descuentos agresivos

**💰 Matriz de Pagos (Miles de pesos)**
**Formato: Utilidad TransCaribe / Utilidad MetroUrbano**

```python
# Definimos la matriz de pagos del problema
# Formato: TransCaribe / MetroUrbano
matriz_pagos = {
    ('A', 'A'): (20, 20),  # Ambos renuevan flota
    ('A', 'B'): (35, 10),  # TC renueva, MU descuentos
    ('B', 'A'): (10, 35),  # TC descuentos, MU renueva
    ('B', 'B'): (25, 25)   # Ambos descuentos
}

# Crear DataFrame para mejor visualizacion
df_matriz = pd.DataFrame({
    'MetroUrbano A': ['20/20', '10/35'],
    'MetroUrbano B': ['35/10', '25/25']
}, index=['TransCaribe A', 'TransCaribe B'])

print("📊 MATRIZ DE PAGOS (Miles de pesos)")
print("="*40)
print(df_matriz)
print("\nInterpretacion:")
print("• (A,A): Ambas renuevan → Beneficios equilibrados")
print("• (A,B): TC domina con renovacion → TC gana mas")  
print("• (B,A): MU domina con renovacion → MU gana mas")
print("• (B,B): Guerra de descuentos → Beneficios moderados")
```

## **II. ANALISIS PREDICTIVO - EQUILIBRIO DE NASH**

### **Analisis de Mejores Respuestas**

**TransCaribe:**
- Si MetroUrbano juega A → TransCaribe elige A (20 > 10) ✅
- Si MetroUrbano juega B → TransCaribe elige A (35 > 25) ✅

**MetroUrbano:**
- Si TransCaribe juega A → MetroUrbano elige A (20 > 10) ✅
- Si TransCaribe juega B → MetroUrbano elige A (35 > 25) ✅

### **Equilibrio de Nash Identificado**
🎯 **(A, A) con utilidades (20, 20)**

**Interpretacion:**
- Ambas empresas tienen **estrategia dominante A** (renovar flota)
- Es un equilibrio **estable**: ninguna empresa quiere desviarse unilateralmente
- Representa **racionalidad individual** de cada empresa

<hr style="
    border: none;
    border-top: 15px solid #00cc8fff;
    border-radius: 8px;
    width: 80%;
    margin: 80px auto;
">

```python
# Analisis de eficiencia de Pareto
resultados = [('A', 'A'), ('A', 'B'), ('B', 'A'), ('B', 'B')]
pareto_optimos = []

# Verificar cada resultado
for resultado in resultados:
    es_pareto = True
    utilidad_actual = matriz_pagos[resultado]
    
    for otro_resultado in resultados:
        if resultado != otro_resultado:
            utilidad_otra = matriz_pagos[otro_resultado]
            # Verificar dominancia
            if (utilidad_otra[0] >= utilidad_actual[0] and utilidad_otra[1] >= utilidad_actual[1] and 
                (utilidad_otra[0] > utilidad_actual[0] or utilidad_otra[1] > utilidad_actual[1])):
                es_pareto = False
                break
    
    if es_pareto:
        pareto_optimos.append((resultado, utilidad_actual))

# Crear tabla de analisis
datos_analisis = {
    'Estrategia': ['(A,A)', '(A,B)', '(B,A)', '(B,B)'],
    'TransCaribe': [20, 35, 10, 25],
    'MetroUrbano': [20, 10, 35, 25],
    'Suma Total': [40, 45, 45, 50],
    'Pareto Optimo': ['❌', '✅', '✅', '✅']
}

df_analisis = pd.DataFrame(datos_analisis)
print("📊 ANALISIS DE EFICIENCIA DE PARETO")
print("="*50)
print(df_analisis.to_string(index=False))

print(f"\n🏆 Resultados Pareto Optimos: {len(pareto_optimos)}")
for resultado, utilidades in pareto_optimos:
    print(f"• {resultado}: {utilidades}")
```

## **III. ANALISIS PRESCRIPTIVO - OPTIMO DE PARETO**

### **Resultados Pareto optimos**
- **(A,B)**: (35,10) - Suma: 45
- **(B,A)**: (10,35) - Suma: 45  
- **(B,B)**: (25,25) - Suma: 50 ⭐

### **Problema que identifique**
❌ **El Equilibrio de Nash (A,A) NO es Pareto optimo**

### **Dilema Nash vs Pareto**
- **Nash (A,A)**: Utilidades (20,20) - Suma: 40
- **optimo Social (B,B)**: Utilidades (25,25) - Suma: 50
- **Pérdida social**: 10 mil pesos por mantenerse en Nash

### **Conflicto que identifique**
Existe tension entre:
- **Racionalidad individual** → (A,A)
- **Eficiencia social** → (B,B)

    border: none;
    border-top: 15px solid #00cc8fff;

```python
# Propuesta de solucion cooperativa
nash_utilidades = (20, 20)
optimo_social = (25, 25)
ganancia_cooperacion = sum(optimo_social) - sum(nash_utilidades)

# Análisis de la propuesta
print("PROPUESTA DE SOLUCION COOPERATIVA")
print("="*45)
print(f"Estrategia recomendada: (B,B) - Ambas ofrecen descuentos")
print(f"Beneficio por empresa: ${optimo_social[0]} mil vs ${nash_utilidades[0]} mil")
print(f"Ganancia adicional: +${optimo_social[0] - nash_utilidades[0]} mil por empresa")
print(f"Beneficio social total: +${ganancia_cooperacion} mil")

# Visualizacion final
fig, ax = plt.subplots(1, 1, figsize=(12, 8))

# Datos para el gráfico
scenarios = ['Equilibrio Nash\n(A,A)', 'Solucion Cooperativa\n(B,B)', 'Ganancia por\nCooperar']
tc_values = [20, 25, 5]
mu_values = [20, 25, 5]
total_values = [40, 50, 10]

x_pos = np.arange(len(scenarios))
width = 0.25

# Crear barras
bars1 = ax.bar(x_pos - width, tc_values, width, label='TransCaribe', 
               color='steelblue', alpha=0.8, edgecolor='black')
bars2 = ax.bar(x_pos, mu_values, width, label='MetroUrbano', 
               color='darkorange', alpha=0.8, edgecolor='black')
bars3 = ax.bar(x_pos + width, total_values, width, label='Suma Total', 
               color='forestgreen', alpha=0.8, edgecolor='black')

ax.set_xlabel('Escenarios de Análisis', fontweight='bold')
ax.set_ylabel('Utilidades (Miles de Pesos)', fontweight='bold')
ax.set_title('Análisis Comparativo: Nash vs Cooperacion\nBeneficios de la Coordinacion Estratégica', 
             fontweight='bold', fontsize=14)
ax.set_xticks(x_pos)
ax.set_xticklabels(scenarios)
ax.legend(loc='upper left')
ax.grid(True, alpha=0.3)

# Añadir valores en las barras
for i, (bar1, bar2, bar3) in enumerate(zip(bars1, bars2, bars3)):
    ax.text(bar1.get_x() + bar1.get_width()/2., bar1.get_height() + 0.3,
             f'{tc_values[i]}', ha='center', fontweight='bold', fontsize=10)
    ax.text(bar2.get_x() + bar2.get_width()/2., bar2.get_height() + 0.3,
             f'{mu_values[i]}', ha='center', fontweight='bold', fontsize=10)
    ax.text(bar3.get_x() + bar3.get_width()/2., bar3.get_height() + 0.3,
             f'{total_values[i]}', ha='center', fontweight='bold', fontsize=10)

plt.tight_layout()
plt.show()
```
