# **Ejercicio de Maximización de Utilidad**

Una empresa desea maximizar su función de utilidad:

$$
\mathbf{U(x, y) = x^{0.4} \cdot y^{0.6}}
$$

**Sujeto a:**

- $$20x + 30y \leq 1200$$
- $$x \geq 0$$
- $$y \geq 0$$

---

### **Preguntas:**

1. **¿Cuál es la combinación óptima de bienes \(x\) e \(y\) que maximiza la utilidad?**
2. **¿Cuál es el valor máximo de la función de utilidad?**
3. **¿Qué sucede si el presupuesto aumenta a 1500?**

---

> 💡 **Resolvamos este problema utilizando métodos de optimización y analicemos cómo cambia la solución al modificar el presupuesto.**

```python
# Funcion objetivo negativa (porque minimize solo minimiza)
def utilidad_neg(x):
    if x[0] <= 0 or x[1] <= 0:
        return 1e10  # penalización
    return - (x[0]**0.4) * (x[1]**0.6)

# Funcion para resolver con un presupuesto dado
def resolver_utilidad(presupuesto):
    def restriccion(x):
        return presupuesto - (20 * x[0] + 30 * x[1])
    
    # Voy a seguir los ejemplos del video de YouTube
    restr = {'type': 'ineq', 'fun': restriccion}
    bounds = [(0, None), (0, None)]
    sol = minimize(utilidad_neg, [10, 10], method='SLSQP', bounds=bounds, constraints=[restr])
    return sol


# Voy a crear dos casos con diferentes presupuestos justamente como se ve en el documento donde se explica el problema

# ****** Caso 1: Presupuesto 1200 ******
sol_1200 = resolver_utilidad(1200)
x1, y1 = sol_1200.x
utilidad1 = -sol_1200.fun

# ****** Caso 2: Presupuesto 1500 ******
sol_1500 = resolver_utilidad(1500)
x2, y2 = sol_1500.x
utilidad2 = -sol_1500.fun

# ****** Resultados ******
print("🔹 Presupuesto: 1200")
print(f"Bien x = {x1:.2f}, Bien y = {y1:.2f}")
print(f"Utilidad máxima = {utilidad1:.4f}\n")

print("🔹 Presupuesto: 1500")
print(f"Bien x = {x2:.2f}, Bien y = {y2:.2f}")
print(f"Utilidad máxima = {utilidad2:.4f}")
```

## **Explicación del código de maximización de utilidad**

1. **Definición de la función objetivo negativa**

   Se define la función `utilidad_neg(x)`, que representa la utilidad a maximizar pero con signo negativo, ya que el método `minimize` de Scipy solo minimiza funciones.  
   - Si alguno de los valores de `x` o `y` es menor o igual a cero, se retorna una penalización muy grande para evitar soluciones no factibles.
   - Si los valores son válidos, retorna el negativo de la función de utilidad:  
     $$
     - (x^{0.4} \cdot y^{0.6})
     $$

2. **Función para resolver el problema con un presupuesto dado**

   - Define la restricción presupuestaria: $20x + 30y \leq \text{presupuesto}$
   - Usa el método SLSQP de `minimize` para encontrar los valores óptimos de \( x \) y \( y \) que maximizan la utilidad, respetando las restricciones y los límites (no negativos).

3. **Resolución para dos presupuestos diferentes**

   - **Caso 1:** Se resuelve el problema con un presupuesto de 1200.
     - Se obtienen los valores óptimos de `x` y `y` y la utilidad máxima alcanzable.
   - **Caso 2:** Se repite el proceso con un presupuesto de 1500.

4. **Impresión de resultados**

   Se muestran los resultados para ambos casos:
   - El valor óptimo de `x` y `y` para cada presupuesto.
   - El valor máximo de la función de utilidad alcanzado en cada caso.

---

**En resumen:**  
Este código permite comparar cómo cambia la combinación óptima de bienes y la utilidad máxima cuando el presupuesto disponible

# **Ejercicio 2: Finanzas - Optimización de Portafolio**

Un inversionista desea asignar su capital entre dos activos para **minimizar el riesgo** de su portafolio. El riesgo se modela con la siguiente función cuadrática:

$$
R(x, y) = 5x^2 + 3y^2 - 2xy
$$

donde:
- \( x \) es la proporción invertida en el activo 1,
- \( y \) es la proporción invertida en el activo 2.

**Restricciones:**
- La suma de las proporciones debe ser igual a 1:
  $$
  x + y = 1
  $$
- No se puede invertir una proporción negativa:
  $$
  x \geq 0,\quad y \geq 0
  $$

---

### **Preguntas a resolver:**

1. **¿Qué proporción de inversión debe destinarse a cada activo para minimizar el riesgo?**
2. **¿Cuál es el riesgo mínimo alcanzable?**
3. **¿Cómo varía la solución si la correlación entre los activos se vuelve más negativa** (es decir, si el término \(-2xy\) cambia a \(-4xy\))?

---

> 💡 **A continuación voy a hacer el código para resolver este problema de optimización y analizar el efecto de una mayor CORRELACION NEGATIVA ENTRE LOS ACTIVOS.**

```python
# Caso 1: -2xy
def riesgo_2xy(x):
    return 5 * x[0]**2 + 3 * x[1]**2 - 2 * x[0] * x[1]

# Caso 2: -4xy (correlacion mas negativa)
def riesgo_4xy(x):
    return 5 * x[0]**2 + 3 * x[1]**2 - 4 * x[0] * x[1]

# Restriccion: x + y = 1
restriccion_eq = {'type': 'eq', 'fun': lambda x: x[0] + x[1] - 1}
bounds = [(0, 1), (0, 1)]
x0 = [0.5, 0.5]

# Solución caso 1
sol1 = minimize(riesgo_2xy, x0, method='SLSQP', bounds=bounds, constraints=[restriccion_eq])
x1, y1 = sol1.x
riesgo_min1 = sol1.fun

# Solución caso 2
sol2 = minimize(riesgo_4xy, x0, method='SLSQP', bounds=bounds, constraints=[restriccion_eq])
x2, y2 = sol2.x
riesgo_min2 = sol2.fun


# Resultados
print("🔹 Caso 1: R(x, y) = 5x2 + 3y2 - 2xy")
print(f"Proporcion optima: x = {x1:.4f}, y = {y1:.4f}")
print(f"Riesgo minimo = {riesgo_min1:.4f}\n")

print("🔹 Caso 2: R(x, y) = 5x2 + 3y2 - 4xy (correlacion mas negativa)")
print(f"Proporcion optima: x = {x2:.4f}, y = {y2:.4f}")
print(f"Riesgo minimo = {riesgo_min2:.4f}")
```

## **Explicación del código de minimización de riesgo**

1. **Definición de las funciones de riesgo**

   - Se definen dos funciones:
     - `riesgo_2xy(x)`: Calcula el riesgo como $$ 5x^2 + 3y^2 - 2xy $$
     - `riesgo_4xy(x)`: Calcula el riesgo como $$ 5x^2 + 3y^2 - 4xy $$ (mayor correlación negativa entre los activos).

2. **Restricción de la cartera**

   - Se impone la restricción de que la suma de las proporciones debe ser 1:  
     $$
     x + y = 1
     $$
   - Además, los valores de \( x \) y \( y \) están acotados entre 0 y 1.

3. **Resolución de ambos casos**

   - Se utiliza el método `minimize` de Scipy para encontrar la combinación óptima de \( `x` \) y \( `y` \) que minimiza el riesgo en cada caso, respetando la restricción y los límites.
   - Se parte de una solución inicial de \( x = 0.5 \), \( y = 0.5 \).

4. **Impresión de resultados**

   - Se muestran los resultados para ambos escenarios:
     - La proporción óptima de cada activo en la cartera.
     - El riesgo mínimo alcanzado en cada caso.

---

**Conclusión:**  
Este análisis permite comparar cómo cambia la asignación óptima y el riesgo mínimo cuando la correlación entre los activos se vuelve más NEGATIVA.

- \( y \) es la cantidad del insumo 2.

**Restricciones:**
- La suma ponderada de insumos no puede superar el recurso disponible:
  $$
  x + 2y \leq 100
  $$
- No se pueden usar cantidades negativas de insumos:
  $$
  x \geq 0,\quad y \geq 0
  $$

---

### **Preguntas a resolver:**

1. **¿Cuál es la combinación óptima de insumos \(x\) e \(y\) para maximizar la producción?**
2. **¿Qué cantidad máxima puede producirse?**
3. **Si se dispone de 150 unidades de recursos, ¿cuál sería el nuevo nivel de producción?**

---

> 💡 **A continuación, el código para resolver este problema de optimización y analizar cómo cambia la producción máxima al AUMENTAR los recursos disponibles.**

```python
# Funcion objetivo negativa (para maximizar)
def produccion_neg(x):
    if x[0] < 0 or x[1] < 0:
        return 1e10
    return -50 * (x[0]**0.5) * (x[1]**0.3)

# Funcion para resolver el problema dado un recurso total
def resolver_produccion(recurso):
    def restriccion(x):
        return recurso - (x[0] + 2*x[1])
    restr = {'type': 'ineq', 'fun': restriccion}
    bounds = [(0, None), (0, None)]
    sol = minimize(produccion_neg, [10, 10], method='SLSQP', bounds=bounds, constraints=[restr])
    return sol

# Caso 1: 100 unidades de recursos
sol_100 = resolver_produccion(100)
x1, y1 = sol_100.x
produccion1 = -sol_100.fun

print("🔹 Recursos: 100")
print(f"Combinación optima: x = {x1:.2f}, y = {y1:.2f}")
print(f"Producción maxima = {produccion1:.4f}\n")

# Caso 2: 150 unidades de recursos
sol_150 = resolver_produccion(150)
x2, y2 = sol_150.x
produccion2 = -sol_150.fun

print("🔹 Recursos: 150")
print(f"Combinación optima: x = {x2:.2f}, y = {y2:.2f}")
print(f"Producción maxima = {produccion2:.4f}")
```

3. **Resolución para dos cantidades de recursos**

   - **Caso 1:** Se resuelve el problema con 100 unidades de recursos.
   - **Caso 2:** Se repite el proceso con 150 unidades de recursos.

Un hotel de playa tiene la siguiente función de ingresos, donde:
- \( x \) es el **precio por noche** (en dólares)
- \( y \) es el **porcentaje de ocupación**

La función de ingresos es:

$$
I(x, y) = x \cdot y - 0.01y^2
$$

**Restricciones:**
- El precio debe estar entre 80 y 200 dólares:
  $$
  80 \leq x \leq 200
  $$
- El porcentaje de ocupación debe estar entre 0% y 100%:
  $$
  0 \leq y \leq 100
  $$

---

### **Preguntas a resolver:**

1. **¿Cuál es el precio óptimo y el nivel de ocupación que maximizan los ingresos?**
2. **¿Cuál es el ingreso máximo que puede alcanzar el hotel?**
3. **¿Qué sucede si se mejora el servicio y la función de ingresos cambia a:**  
   $$
   I(x, y) = x \cdot y - 0.005y^2
   $$

---

> 💡 **A continuación se presenta el código para resolver este problema de optimización y analizar el efecto de una mejora en el servicio sobre los ingresos máximos del hotel**

```python
# Escenario 1: I(x, y) = x*y - 0.01*y^2
def ingresos1(x):
    return - (x[0]*x[1] - 0.01*x[1]**2)

bounds = [(80, 200), (0, 100)]
sol1 = minimize(ingresos1, [100, 50], method='SLSQP', bounds=bounds)
x1, y1 = sol1.x
ingreso_max1 = -sol1.fun

print("🔹 Escenario 1: I(x, y) = x*y - 0.01*y^2")
print(f"Precio optimo: ${x1:.2f}")
print(f"Ocupacion optima: {y1:.2f}%")
print(f"Ingreso maximo: ${ingreso_max1:.2f}\n")

# Escenario 2: I(x, y) = x*y - 0.005*y^2 (mejora del servicio)
def ingresos2(x):
    return - (x[0]*x[1] - 0.005*x[1]**2)

sol2 = minimize(ingresos2, [100, 50], method='SLSQP', bounds=bounds)
x2, y2 = sol2.x
ingreso_max2 = -sol2.fun

print("🔹 Escenario 2: I(x, y) = x*y - 0.005*y^2 (mejora del servicio)")
print(f"Precio optimo: ${x2:.2f}")
print(f"Ocupacion optima: {y2:.2f}%")
print(f"Ingreso maximo: ${ingreso_max2:.2f}")
```

# **Ejercicio 5: Educación - Optimización de Recursos en una Escuela**

Una escuela busca asignar recursos entre **capacitación docente** (\(x\)) y **material educativo** (\(y\)) para maximizar el rendimiento académico de los estudiantes, según la siguiente función:

$$
R(x, y) = 2x^{0.5} + 3y^{0.5}
$$

**Restricciones:**
- El presupuesto total debe cumplir:
  $$
  50x + 30y \leq 3000
  $$
- No se pueden asignar cantidades negativas:
  $$
  x \geq 0,\quad y \geq 0
  $$

---

### **Preguntas a resolver:**

1. **¿Cuál es la asignación óptima de recursos en capacitación y materiales?**
2. **¿Cuál es el rendimiento académico esperado?**
3. **Si el presupuesto aumenta a 4000, ¿cómo afecta la asignación óptima?**

---

> 💡 **A continuación se presenta el código para resolver este problema de optimización y analizar cómo cambia la asignación óptima y el rendimiento académico al aumentar el presupuesto.**

```python
def objetivo(x):
    if x[0] < 0 or x[1] < 0:
        return 1e10
    return - (2 * x[0]**0.5 + 3 * x[1]**0.5)

def restriccion(presupuesto):
    return {'type': 'ineq', 'fun': lambda x: presupuesto - (50*x[0] + 30*x[1])}

bounds = [(0, None), (0, None)]

# Caso 1: Presupuesto 3000
sol_3000 = minimize(objetivo, [10, 10], method='SLSQP', bounds=bounds, constraints=[restriccion(3000)])
x1, y1 = sol_3000.x
rendimiento1 = -sol_3000.fun

print("🔹 Presupuesto: 3000")
print(f"Capacitación docente óptima: x = {x1:.2f}")
print(f"Material educativo óptimo: y = {y1:.2f}")
print(f"Rendimiento académico esperado = {rendimiento1:.4f}\n")

# Caso 2: Presupuesto 4000
sol_4000 = minimize(objetivo, [10, 10], method='SLSQP', bounds=bounds, constraints=[restriccion(4000)])
x2, y2 = sol_4000.x
rendimiento2 = -sol_4000.fun

print("🔹 Presupuesto: 4000")
print(f"Capacitación docente óptima: x = {x2:.2f}")
print(f"Material educativo óptimo: y = {y2:.2f}")
print(f"Rendimiento académico esperado = {rendimiento2:.4f}")
```

## **Explicación del código de maximización de rendimiento académico**

1. **Definición de la función objetivo negativa**

   - Se define la función `objetivo(x)`, que representa el rendimiento académico a maximizar, pero con signo negativo (porque `minimize` solo minimiza).
   - Si alguno de los valores de \( x \) o \( y \) es negativo, retorna una penalización grande.
   - Si los valores son válidos, retorna:
     $$
     - \left( 2 \sqrt{x} + 3 \sqrt{y} \right)
     $$

2. **Definición de la restricción presupuestaria**

   - Se crea la función `restriccion(presupuesto)` que define la restricción:
     $$
     50x + 30y \leq \text{presupuesto}
     $$
   - Esto asegura que la suma del gasto en capacitación docente y material educativo no supere el presupuesto disponible.

3. **Resolución para dos presupuestos diferentes**

   - **Caso 1:** Se resuelve el problema con un presupuesto de 3000.
   - **Caso 2:** Se repite el proceso con un presupuesto de 4000.

4. **Impresión de resultados**

   - Se muestran los resultados para ambos casos:
     - La cantidad óptima de capacitación docente (\( x \)).
     - La cantidad óptima de material educativo (\( y \)).
     - El rendimiento académico máximo esperado.
